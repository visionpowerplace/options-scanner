## 3. Your settingsChange these two numbers to match your account, then click ▶.- `ACCOUNT` — your account size in dollars- `RISK` — percent of the account you are willing to lose on ONE tradeOn a small account `RISK` has to be high for anything to be affordable, and that is a realrisk, not a formality: at 20%, five losing trades in a row costs you about three quartersof the account. Five in a row is an ordinary run.

```python
ACCOUNT = 500      # your account size in dollarsRISK    = 20       # percent of account risked per tradeCONFIG  = "micro-500.yaml"   # set to "" to use the standard settings instead# Which symbols to scan. "liquid" is the tightest-quoted ~70 names.# Other options: "small", "etf", "mega", "default", or your own list like "SPY,QQQ,AAPL"UNIVERSE = "liquid"print(f"Account ${ACCOUNT:,} | risking {RISK}% = ${ACCOUNT*RISK/100:,.0f} per trade")
```

## 4. Try it on simulated data firstThis uses a built-in market simulator — no live data, nothing tradeable. It is here so youcan see the whole output format and check your settings before pointing it at the realmarket. Click ▶.

## 5. The real scanLive market data. Run this **after the US market closes** (after 3pm Central / 4pm Eastern),or before it opens. End-of-day quotes are settled; the tool is built on daily bars.Click ▶. This takes 1–3 minutes depending on how many symbols you scan.

```python
import subprocess, syscmd = [sys.executable, "-m", "optionscan", "scan", "--provider", "yahoo",       "--universe", UNIVERSE, "--account", str(ACCOUNT), "--risk", str(RISK), "--quiet"]if CONFIG: cmd += ["--config", CONFIG]r = subprocess.run(cmd, capture_output=True, text=True)print(r.stdout[:20000])if r.returncode != 0: print("STDERR:", r.stderr[-3000:])
```
